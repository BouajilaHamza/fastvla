## 3. Load Model in 4-bit (QLoRA)
FastVLA loads large models with 4-bit quantization. The `action_dim` parameter must align with your specific robotics dataset (e.g., 2 for PushT, 7 for standard arms).

**Efficiency Tip:** For immediate access and high-efficiency training, use `"smolvla"`. For full reasoning capabilities, use `"openvla-7b"` (FastVLA provides 15x higher parameter-efficiency for 7B models).

```python
from fastvla import FastVLAModel
import torch

# ── FastVLA Flagship Moat: Fine-Tune 7B Foundation Models on Free 16GB GPU (T4) ──
# • "openvla-7b": Flagship 7B Robotics Model (requires Meta Llama-2 access via HF_TOKEN)
# • "smolvla": Fast ungated fallback if Meta approval is pending
model_id = "openvla-7b"  # Change to "smolvla" if you don't have Llama-2 access yet

# Robot Action & State dimensions (PushT = 2, LIBERO = 7, Franka = 7)
ACTION_DIM = 2

print(f"Loading {model_id} in 4-bit NF4...")
model = FastVLAModel.from_pretrained(
    model_id,
    load_in_4bit=True,
    use_peft=True,
    gradient_checkpointing=True,
    action_dim=ACTION_DIM,
    state_dim=ACTION_DIM,
)

print(f"\n✓ Model loaded successfully.")
if torch.cuda.is_available():
    print(f"Current Peak VRAM Allocation: {torch.cuda.memory_allocated() / 1024**3:.2f} GB")
```

## 4. Fine-Tuning on PushT
Training involves the real-world block pushing dataset. FastVLA automatically manages image resolution interpolation (e.g., 224px to 384px) and gradient accumulation.

**Production Features:**
- **Automated Push-to-Hub**: Save and deploy your trained adapters directly to Hugging Face.
- **Dynamic Resizing**: Seamless handling of dataset-model resolution mismatches.
- **Memory-Optimized**: Integrated 8-bit optimizers for low-latency T4 training.

```python
from fastvla import FastVLATrainer
from fastvla.data.datasets import get_dataset
import torch

# Load production dataset (25,650 Push-T manipulation frames)
dataset = get_dataset("lerobot/pusht_image")

# ── High-Throughput Configuration Utilizing 16GB T4 VRAM Headroom ──
# • batch_size=16 saturates Tensor Cores and resolves Colab CPU dataloader bottlenecks
# • max_steps=100 provides quick fine-tuning convergence without running all 6,413 batches
trainer = FastVLATrainer(
    model=model,
    dataset=dataset,
    batch_size=16,                  # High throughput (saturates VRAM)
    gradient_accumulation_steps=1,
    lr=2e-4,
    max_steps=100,                  # Set to None or num_epochs=1 to run full 6,413 batches
    num_workers=2,                  # Matches Colab 2 vCPUs
    pin_memory=True,
    use_mixed_precision=True,
    use_8bit_optimizer=True,
    output_dir="./checkpoints",
    logging_steps=20,
)

print("Starting Training Loop...")
trainer.train()
print("\n✓ Training complete!")

# ── Hardware & Telemetry Profiling ──
if torch.cuda.is_available():
    allocated = torch.cuda.memory_allocated() / (1024**3)
    max_alloc = torch.cuda.max_memory_allocated() / (1024**3)
    print(f"\n📊 Colab T4 Performance Profile:")
    print(f"  • Active VRAM:     {allocated:.2f} GB / 15.00 GB")
    print(f"  • Peak VRAM:       {max_alloc:.2f} GB ({max_alloc/15.0*100:.1f}% capacity)")
    print(f"  • Free Headroom:   {15.0 - max_alloc:.2f} GB")
```

## 6. Inference (Predict Action)
Execute an end-to-end forward pass to predict the next robot action based on a visual observation.

```python
import numpy as np
import torch
from PIL import Image
from torchvision import transforms

dummy_image = Image.fromarray(np.random.randint(0, 255, (224, 224, 3), dtype=np.uint8))
prompt = "push the t-shaped block into the goal zone"

# FastVLA automatically handles image transformations
transform = transforms.Compose([
    transforms.Resize(model.config.image_size),
    transforms.ToTensor(),
])

device = model.device
pixel_values = transform(dummy_image).unsqueeze(0).unsqueeze(0).to(device)
input_ids = model.tokenizer(prompt, return_tensors="pt")["input_ids"].to(device)

model.eval()
# Use autocast to seamlessly match 4-bit / fp16 compute dtypes
with torch.no_grad():
    if torch.cuda.is_available():
        with torch.autocast(device_type="cuda", dtype=torch.float16):
            action, _ = model(pixel_values=pixel_values, input_ids=input_ids)
    else:
        action, _ = model(pixel_values=pixel_values, input_ids=input_ids)

print(f"✓ Inference Succeeded!")
print(f"Predicted Action Shape: {action.shape}  (Batch, Chunk_Size, Action_Dim)")
print(f"Action Waypoints:")
print(action[0, :min(4, action.shape[1])].cpu().numpy())
```
